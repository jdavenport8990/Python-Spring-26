```python
import sys

class Task:
    def __init__(self, name, priority):
        self.name = name
        self.priority = priority

    def __str__(self):
        labels = {1: "🔴 High", 2: "🟡 Medium", 3: "🟢 Low"}
        priority_label = labels.get(self.priority, "Unknown")
        return f"[{priority_label}] {self.name}"

class TaskManager:
    def __init__(self):
        self.tasks = []
        self.history_file = "task_history.txt"

    def add_task(self):
        name = input("\nEnter task name: ").strip()
        try:
            priority = int(input("Enter priority (1-High, 2-Med, 3-Low): "))
            if priority not in [1, 2, 3]: raise ValueError
            self.tasks.append(Task(name, priority))
            print(f"✅ Added: {name}")
        except ValueError:
            print("❌ Invalid priority.")

    def show_tasks(self):
        print("\n--- Active Tasks ---")
        if not self.tasks:
            print("No active tasks.")
        else:
            for i, task in enumerate(self.tasks, 1):
                print(f"{i}. {task}")
        print("--------------------")

    def complete_task(self):
        self.show_tasks()
        if not self.tasks:
            return

        try:
            choice = int(input("Enter the task number to mark as done: "))
            # Remove task from list (adjusting for 1-based indexing)
            completed_task = self.tasks.pop(choice - 1)

            # Save to permanent file
            with open(self.history_file, "a") as file:
                file.write(f"COMPLETED: {completed_task.name}\n")

            print(f"🎉 Task '{completed_task.name}' completed and saved to history!")
        except (ValueError, IndexError):
            print("❌ Invalid task number.")

def main():
    manager = TaskManager()

    while True:
        print("\n1. Add Task\n2. View Tasks\n3. Complete Task\n4. Exit")
        choice = input("Select an option: ")

        if choice == "1":
            manager.add_task()
        elif choice == "2":
            manager.show_tasks()
        elif choice == "3":
            manager.complete_task()
        elif choice == "4":
            print("Goodbye!")
            break
        else:
            print("Invalid choice.")

if __name__ == "__main__":
    main( )
```
