```python
import sys
class Task:
    """Represents a single task with a name and priority."""
    def __init__(self, name, priority):
        self.name = name
        self.priority = priority

    def __str__(self):
        # Maps the integer to a readable label with icons for clarity
        labels = {1: "🔴 High", 2: "🟡 Medium", 3: "🟢 Low"}
        priority_label = labels.get(self.priority, "Unknown")
        return f"[{priority_label}] {self.name}"

class TaskManager:
    """Handles the storage, sorting, and display of tasks."""
    def __init__(self):
        self.tasks = []

    def add_task(self):
        name = input("\nEnter task name: ").strip()
        if not name:
            print("❌ Task name cannot be empty.")
            return

        try:
            priority = int(input("Enter priority (1-High, 2-Med, 3-Low): "))
            if priority not in [1, 2, 3]:
                raise ValueError

            new_task = Task(name, priority)
            self.tasks.append(new_task)
            print(f"✅ Added: {name}")
        except ValueError:
            print("❌ Invalid input. Please enter 1, 2, or 3 for the priority.")

    def show_tasks(self):
        print("\n--- Current Task List ---")
        if not self.tasks:
            print("Your list is currently empty.")
        else:
            # Sorts by priority (1 before 2, etc.) so High priority stays at the top
            sorted_tasks = sorted(self.tasks, key=lambda x: x.priority)
            for i, task in enumerate(sorted_tasks, 1):
                print(f"{i}. {task}")
        print("-------------------------\n")

def main():
    manager = TaskManager()

    # Mapping choices to functions keeps the main loop clean
    menu_options = {
        "1": manager.add_task,
        "2": manager.show_tasks,
        "3": sys.exit
    }

    while True:
        print("Personal Task Manager")
        print("1. Add Task")
        print("2. View Tasks")
        print("3. Exit")

        choice = input("Select an option: ")

        if choice in menu_options:
            if choice == "3":
                print("Exiting... Happy hacking!")
                menu_options[choice]()
            menu_options[choice]()
        else:
            print("\n⚠️  Invalid choice, please select 1, 2, or 3.\n")

if __name__ == "__main__":
    main()
```
