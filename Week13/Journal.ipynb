```python
from os import read
# NAME: journal.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS

#Function 1: write_entry(filename, entry)
def write_entry(filename, entry):

#Must be opened in append mode ('a')
    file = open(filename, 'a')

#Write the entry to the file, new line for a new entry ('/n')
    file.write(entry + '\n')

#Prompt to close file
    file.close()

#Function 2: read_journal(filename)
def read_journal(filename):
    try:
#Must be opened in ('r')
       file = open(filename, 'r')

#Read the entire contents
       contents = file.read()
       file.close()

#Print contents for user
       print("\n*** Your Journal Entries ***\n")
       print(contents)
    except FileNotFoundError:
        print("\n No entries found. \n")



# PROGRAM LOGIC

filename = "journal.txt"

#Ask the user if they want to (W)rite a new entry, (R)ead their journal or (Q)uit.
while True:
    choice = input("Do you want to (W)rite a new entry, (R)ead their journal or (Q)uit?").lower()

#If they choose 'W', ask for their thoughts and call write_entry.
    if choice == "w":
        entry = input("What's on your mind?: ")
        write_entry(filename, entry)
        print("Your entry has been saved.")

#If they choose 'R', call read_journal.
    elif choice == "r":
        read_journal(filename)

#Wrap this in a while loop so the user can keep using app unitl Q
    elif choice == "q":
        print("Goodbye!")
        break
```
