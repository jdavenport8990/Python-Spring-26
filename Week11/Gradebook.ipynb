```python
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS
#Function 1 add_students
def add_students(gradebook):
#accepts a dictionary as a parameter
#While loop:
    while True:
#Ask for student name
        name = input("Enter student name (or 'done'): ")
#if name == "done": break
        if name.lower() == "done":
            break
#Ask for grade
        grade = int(input(f"Enter {name}'s grade: "))
#Add student and grade to dictionary
        gradebook[name] = grade


#Function 2 calculate_average
def calculate_average(gradebook):
    if len(gradebook) == 0:
        return none
#accepts a dictionary as a parameter

#get all the values from a dictionary using .values()
    grades = gradebook.values()

#Calculate average (sum / count)
    average = sum(grades) / len(grades)
#Return the average
    return average




# PROGRAM LOGIC
#Initialize an empty dictionary called gradebook.
gradebook = {}
#Call add_students to populate the dictionary.
add_students(gradebook)

#Call calculate_average and display the resulting class average.
average = calculate_average(gradebook)

#Print a final report showing each student's name and their grade
print("---Class Report---")

for name, grade in gradebook.items():
    print(f"{name}: {grade}")
if average is None:
    print(f"Class average: {N/A}")
else:
    print(f"Class average: {average: .2f}")
```
