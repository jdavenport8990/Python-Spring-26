```python
# NAME: roster.py
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS
#Function 1 (add_players):
def add_players(roster_list):

#While loop, if true ask for a players name
    while True:
        name = input("Enter a player's name (type 'quit' to stop): ")
#if user types quit then stop
        if name.lower() == "quit":
            break
#otherwise append name to the list
        else:
            roster_list.append(name)


#Function 2 (is_on_roster) (roster_list, name)
def is_on_roster(roster_list, name):
    for player in roster_list:
#if player matches the name return true
      if player.lower() == name.lower():
        return True
#otherwise return false (if no match)
    else:
        return False

# PROGRAM LOGIC

#Create an empty list called roster
roster = []

#Call add_players to fill the list(roster)
add_players(roster)

#Ask the user for one specific name to search for
search_name = input("Enter a name to search for: ")

#Call is_on_roster(roster, search_name)
if is_on_roster(roster, search_name):

# If true print "Player found!"
    print("Player found!")
#else print "Player not found."
else:
    print("Player not found.")
```
