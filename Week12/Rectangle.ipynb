```python
# NAME: rectangle.py
# IMPORT STATEMENTS #

# FUNCTION DEFINITIONS #

#Define function calculate_rectangle(length, width)
def calculate_rectangle(length, width):

#Calculate area(length * width)
    area = length * width

#Calculate perimeter(length*width) * 2
    perimeter = (length * width) * 2

#Return area and perimeter
    return area, perimeter

#Return both values as a tuple



# PROGRAM LOGIC #

#Ask the user to input a length and a width (convert these to floats).

#Call your calculate_rectangle function using the user's input.
print(calculate_rectangle(8, 12))

#Unpack the returned tuple into two separate variables called my_area and my_perimeter.

#Print the results clearly for the user.
```
