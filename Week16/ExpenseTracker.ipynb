```python
# IMPORT STATEMENTS

# FUNCTION DEFINITIONS

#Function 1 add_expense(expenses_list):
def add_expense(expenses_list):
#Ask for category
    category = input("Enter category: ")

#Ask for vendor
    vendor = input("Enter vendor: ")

#Ask for amount
    while True:
        amount_input = input("Enter amount:")

#repeat until number is entered by user
        try:
            amount = float(amount_input)
            break
        except ValueError:
            print("Invalid input. Please enter a number.")

#Ask for a date
    date = input("Enter date: ")

#Expenses will be a dictionary with keys
#category#vendor#Amount#date
    expense = {
        "category": category,
        "vendor": vendor,
        "amount": amount,
        "date": date
    }

#Append dictionary to expense_list
    expenses_list.append(expense)
    print("Expense added!")

#Function 2 total_all (expense_list)
def total_all(expense_list):
#Set total to = 0
    total = 0
#For each expense in expense_list:
    for expense in expense_list:
#add expense ("amount") to total
        total += expense["amount"]
#return total
    return total

#Function 3 total_category(expense_list, category)
def total_category(expense_list, category):
#set total to = 0
    total = 0
    category = category.lower()
#if expense ("category") matches category:
    for expense in expense_list:
        if expense["category"].lower == category:
#add expense (amount) to total
            total += expense["amount"]
#return total
    return total


#Function 4 total_vendor (expense_list, vendor)
def total_vendor(expense_list, vendor):
#set total to = 0
    total = 0
    vendor = vendor.lower()
#if expense ("vendor") matches vendor:
    for expense in expense_list:
        if expense["vendor"].lower == vendor:

#add expense (amount) to total
            total += expense["amount"]
#return total
    return total

#Function 5 show_totals_menu (expense_list):
def show_totals_menu(expense_list):

#Display options: All, Category, Vendor
    print("\n*** Totals ***")
    print("A: All Expenses")
    print("C: Category")
    print("V: Vendor")

#User enter choice
    choice = input("Enter your choice: ").lower()

#If choice is All:
#Display totals formatted as money
    if choice == "A":
        total = total_all(expense_list)
#Call total_all
        print(f"Total of All Expenses: ${total:.2f}")
#If choice is Category:
    elif choice == "C":
        category = input("Enter category: ")
#ask user for category call total_category
        total = total_category(expense_list, category)
        print(f"Total for category '{category}': ${total:.2f}")

#If choice is Vendor:
    elif choice == "V":
        vendor = input("Enter vendor: ")
        total = total_vendor(expense_list, vendor)
#ask user for vendor call total_vendor
        print(f"Total for vendor '{vendor}': ${total:.2f}")



# PROGRAM LOGIC

#Create a list called expenses loop indefinitely
expenses = []
while True:
#Display Main Menu - 1. Add New Expense, 2. Get Totals, 3. Exit
    print("\n*** Expense Tracker ***")
    print("1. Add New Expense")
    print("2. Get Totals")
    print("3. Exit")

#Obtain user selection -
    selection = input("Enter your selection: ")
#If selection is 1: call(add_expense) expenses
    if selection == "1":
        add_expense(expenses)
        print(expenses)
#If  selection is 2: call show_totals_menu(expenses)
    elif selection == "2":
        show_totals_menu(expenses)
#If selection is 3: print goodbye message
    elif selection == "3":
#Break loop
        print("Goodbye!")
        break
#Else:
#print "Invalid selection
    else:
        print("Invalid selection. Please try again.")
```
